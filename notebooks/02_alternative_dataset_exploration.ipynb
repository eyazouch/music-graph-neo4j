**Chargement**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Charger le nouveau dataset
df_new = pd.read_csv('../data/raw/dataset.csv')

print("=" * 80)
print("NOUVEAU DATASET - APERÇU")
print("=" * 80)

print(f"\n📊 Lignes : {len(df_new):,}")
print(f"📊 Colonnes : {len(df_new.columns)}")

print("\n📋 Colonnes disponibles :")
for i, col in enumerate(df_new.columns, 1):
    print(f"  {i}. {col}")

print("\n📋 Premières lignes :")
df_new.head()
```

**Statistiques rapides**

```python
print("=" * 80)
print("STATISTIQUES RAPIDES")
print("=" * 80)

# Artistes
if 'artists' in df_new.columns:
    print(f"\n🎤 Artistes uniques : {df_new['artists'].nunique():,}")
    print(f"🏆 Top 5 artistes :")
    print(df_new['artists'].value_counts().head(5))

# Années (chercher colonne année)
year_cols = [col for col in df_new.columns if 'year' in col.lower() or 'date' in col.lower()]
print(f"\n📅 Colonnes temporelles : {year_cols}")

if year_cols:
    year_col = year_cols[0]
    print(f"\n📊 Années : {df_new[year_col].min()} à {df_new[year_col].max()}")

# Échantillon
print("\n📋 Échantillon aléatoire :")
df_new.sample(5)
```

**Analyse des genres**

```python
print("=" * 80)
print("ANALYSE DES GENRES")
print("=" * 80)

print(f"\n🎵 Genres uniques : {df_new['track_genre'].nunique()}")

print(f"\n🏆 Top 20 genres :")
top_genres = df_new['track_genre'].value_counts().head(20)
for i, (genre, count) in enumerate(top_genres.items(), 1):
    print(f"  {i:2d}. {genre:20s} : {count:,} tracks")
```

**Analyse de la popularité**

```python
print("=" * 80)
print("ANALYSE DE LA POPULARITÉ")
print("=" * 80)

print(f"\n📊 Statistiques popularité :")
print(f"  - Min : {df_new['popularity'].min()}")
print(f"  - Max : {df_new['popularity'].max()}")
print(f"  - Moyenne : {df_new['popularity'].mean():.1f}")
print(f"  - Médiane : {df_new['popularity'].median()}")

print(f"\n📊 Distribution :")
print(f"  - Popularité 0-20 : {((df_new['popularity'] >= 0) & (df_new['popularity'] <= 20)).sum():,} tracks")
print(f"  - Popularité 21-40 : {((df_new['popularity'] >= 21) & (df_new['popularity'] <= 40)).sum():,} tracks")
print(f"  - Popularité 41-60 : {((df_new['popularity'] >= 41) & (df_new['popularity'] <= 60)).sum():,} tracks")
print(f"  - Popularité 61-80 : {((df_new['popularity'] >= 61) & (df_new['popularity'] <= 80)).sum():,} tracks")
print(f"  - Popularité 81-100 : {(df_new['popularity'] >= 81).sum():,} tracks")

print(f"\n💡 Tracks populaires (>60) : {(df_new['popularity'] > 60).sum():,}")
```

**Distribution artistes**

```python
print("=" * 80)
print("DISTRIBUTION DES ARTISTES")
print("=" * 80)

tracks_per_artist = df_new['artists'].value_counts()

print(f"\n📊 Statistiques tracks par artiste :")
print(f"  - Min : {tracks_per_artist.min()}")
print(f"  - Max : {tracks_per_artist.max()}")
print(f"  - Médiane : {tracks_per_artist.median()}")
print(f"  - Moyenne : {tracks_per_artist.mean():.1f}")

print(f"\n🎯 Artistes avec :")
print(f"  - 1 track : {(tracks_per_artist == 1).sum():,}")
print(f"  - 2-5 tracks : {((tracks_per_artist >= 2) & (tracks_per_artist <= 5)).sum():,}")
print(f"  - 6-10 tracks : {((tracks_per_artist >= 6) & (tracks_per_artist <= 10)).sum():,}")
print(f"  - 11+ tracks : {(tracks_per_artist >= 11).sum():,}")

print(f"\n💡 Si on garde artistes avec 3+ tracks :")
artists_3plus = tracks_per_artist[tracks_per_artist >= 3]
print(f"  - Nombre d'artistes : {len(artists_3plus):,}")
print(f"  - Nombre de tracks : {artists_3plus.sum():,}")
```

**Simulation filtrage popularité**

```python
print("=" * 80)
print("SIMULATION FILTRAGE PAR POPULARITÉ")
print("=" * 80)

thresholds = [30, 40, 50, 60, 70]

for threshold in thresholds:
    filtered = df_new[df_new['popularity'] >= threshold]
    n_artists = filtered['artists'].nunique()
    n_tracks = len(filtered)
    
    print(f"\n🎯 Popularité >= {threshold}")
    print(f"  📊 Tracks : {n_tracks:,}")
    print(f"  🎤 Artistes uniques : {n_artists:,}")
    print(f"  📈 % du dataset : {(n_tracks/len(df_new)*100):.1f}%")
```

**NETTOYAGE COMPLET (SANS FILTRAGE)**


```python
print("=" * 80)
print("NETTOYAGE MINIMAL - TOUS LES ARTISTES")
print("=" * 80)

# Dataset complet
df_cleaned = df_new.copy()

print(f"\n📊 Dataset de départ : {len(df_cleaned):,} tracks")
print(f"🎤 Artistes : {df_cleaned['artists'].nunique():,}")

# 1. Supprimer "Various Artists"
print("\n📊 ÉTAPE 1 : Suppression 'Various Artists'")
various_count = df_cleaned['artists'].str.contains('Various Artists', case=False, na=False).sum()
df_cleaned = df_cleaned[~df_cleaned['artists'].str.contains('Various Artists', case=False, na=False)]
print(f"🗑️ Tracks supprimées : {various_count:,}")

# 2. Supprimer valeurs manquantes (critiques uniquement)
print("\n📊 ÉTAPE 2 : Suppression valeurs manquantes")
before = len(df_cleaned)
df_cleaned = df_cleaned.dropna(subset=['artists', 'track_name'])
print(f"🗑️ Tracks supprimées : {before - len(df_cleaned)}")

# 3. Normalisation
print("\n📊 ÉTAPE 3 : Normalisation")
df_cleaned['artists'] = df_cleaned['artists'].str.strip()
df_cleaned['track_name'] = df_cleaned['track_name'].str.strip()
df_cleaned['album_name'] = df_cleaned['album_name'].str.strip()

print("\n" + "=" * 80)
print("✅ NETTOYAGE TERMINÉ")
print("=" * 80)
print(f"\n📊 Résultat final :")
print(f"  - Tracks : {len(df_cleaned):,}")
print(f"  - Artistes uniques : {df_cleaned['artists'].nunique():,}")
print(f"  - Genres uniques : {df_cleaned['track_genre'].nunique():,}")
print(f"  - Tracks par artiste (médiane) : {df_cleaned['artists'].value_counts().median()}")
```

**EXPORT FINAL**

```python
print("=" * 80)
print("EXPORT DU DATASET COMPLET")
print("=" * 80)

# Export XLSX principal
xlsx_path = '../data/processed/spotify_tracks_complete.xlsx'
df_cleaned.to_excel(xlsx_path, index=False, engine='openpyxl')
print(f"✅ XLSX exporté : {xlsx_path}")
print(f"📊 Taille : {len(df_cleaned):,} lignes × {len(df_cleaned.columns)} colonnes")

# Export CSV (backup)
csv_path = '../data/processed/spotify_tracks_complete.csv'
df_cleaned.to_csv(csv_path, index=False)
print(f"✅ CSV exporté : {csv_path}")

# Liste unique d'artistes
artists_df = df_cleaned[['artists']].drop_duplicates().sort_values('artists')
artists_df.to_excel('../data/processed/artists_complete.xlsx', index=False, engine='openpyxl')
artists_df.to_csv('../data/processed/artists_complete.csv', index=False)
print(f"\n✅ Liste d'artistes : {len(artists_df):,} artistes uniques")

print("\n" + "=" * 80)
print("✅ TOUS LES FICHIERS EXPORTÉS")
print("=" * 80)
print("\n📁 Fichiers créés :")
print("  1. spotify_tracks_complete.xlsx")
print("  2. spotify_tracks_complete.csv")
print("  3. artists_complete.xlsx")
print("  4. artists_complete.csv")
```
