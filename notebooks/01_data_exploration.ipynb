**Import et chargement**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Charger le dataset
df = pd.read_csv('../data/raw/tracks_features.csv')

# Afficher les premières lignes
print("=" * 80)
print("PREMIÈRES LIGNES DU DATASET")
print("=" * 80)
df.head(10)
```

**Informations générales**

```python
print("=" * 80)
print("INFORMATIONS GÉNÉRALES")
print("=" * 80)

print(f"\n📊 Nombre total de lignes : {len(df):,}")
print(f"📊 Nombre total de colonnes : {len(df.columns)}")

print("\n📋 Liste des colonnes :")
for i, col in enumerate(df.columns, 1):
    print(f"  {i}. {col}")

print("\n🔢 Types de données :")
print(df.dtypes)

print("\n❓ Valeurs manquantes :")
missing = df.isnull().sum()
missing_percent = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'Missing': missing[missing > 0],
    'Percent': missing_percent[missing > 0]
})
if len(missing_df) > 0:
    print(missing_df)
else:
    print("✅ Aucune valeur manquante!")
```

**Statistiques artistes**

```python
print("=" * 80)
print("STATISTIQUES ARTISTES")
print("=" * 80)

# Nombre d'artistes uniques
n_artists = df['artists'].nunique()
print(f"\n🎤 Nombre d'artistes uniques : {n_artists:,}")

# Top 10 artistes
print("\n🏆 Top 10 artistes (par nombre de tracks) :")
top_artists = df['artists'].value_counts().head(10)
for i, (artist, count) in enumerate(top_artists.items(), 1):
    print(f"  {i}. {artist}: {count} tracks")
```

**Échantillon de données**

```python
print("=" * 80)
print("ÉCHANTILLON ALÉATOIRE (5 TRACKS)")
print("=" * 80)
df.sample(5)
```

```python
print("=" * 80)
print("DISTRIBUTION DES ARTISTES")
print("=" * 80)

tracks_per_artist = df['artists'].value_counts()

print(f"\n📊 Statistiques tracks par artiste :")
print(f"  - Minimum : {tracks_per_artist.min()}")
print(f"  - Maximum : {tracks_per_artist.max()}")
print(f"  - Médiane : {tracks_per_artist.median()}")
print(f"  - Moyenne : {tracks_per_artist.mean():.1f}")

print(f"\n🎯 Artistes avec :")
print(f"  - 1 track : {(tracks_per_artist == 1).sum():,}")
print(f"  - 2-5 tracks : {((tracks_per_artist >= 2) & (tracks_per_artist <= 5)).sum():,}")
print(f"  - 6-10 tracks : {((tracks_per_artist >= 6) & (tracks_per_artist <= 10)).sum():,}")
print(f"  - 11+ tracks : {(tracks_per_artist >= 11).sum():,}")

print(f"\n💡 Si on garde les artistes avec 5+ tracks :")
artists_5plus = tracks_per_artist[tracks_per_artist >= 5]
print(f"  - Nombre d'artistes : {len(artists_5plus):,}")
print(f"  - Nombre de tracks : {artists_5plus.sum():,}")
```

**Trouver le bon seuil**

```python
print("=" * 80)
print("TROUVER LE SEUIL POUR 3000 ARTISTES")
print("=" * 80)

tracks_per_artist = df['artists'].value_counts()

# Test différents seuils pour approcher 3000
for threshold in [8, 10, 12, 15, 18, 20]:
    n_artists = (tracks_per_artist >= threshold).sum()
    print(f"Seuil {threshold}+ tracks → {n_artists:,} artistes")

# Ou prendre directement le top 3000
print("\n" + "=" * 80)
print("OU SIMPLEMENT : TOP 3000 artistes (peu importe nb tracks)")
top_3000 = tracks_per_artist.head(3000)
print(f"✅ Top 3000 artistes")
print(f"📊 Total tracks : {top_3000.sum():,}")
print(f"📈 % du dataset : {(top_3000.sum()/len(df)*100):.1f}%")
```

**🧹 ÉTAPE 2.5 : Nettoyage et Filtrage des Données**

 **Objectif**
 


**Suppression "Various Artists" (AVANT filtrage)**

```python
print("=" * 80)
print("SUPPRESSION 'VARIOUS ARTISTS' - AVANT FILTRAGE")
print("=" * 80)

print(f"\n📊 Dataset complet avant nettoyage : {len(df):,} tracks")
print(f"📊 Artistes uniques : {df['artists'].nunique():,}")

# Compter combien de tracks ont "Various Artists"
various_count = df['artists'].str.contains('Various Artists', case=False, na=False).sum()
print(f"\n⚠️ Tracks avec 'Various Artists' : {various_count:,}")

# Supprimer toutes les lignes avec "Various Artists"
df = df[~df['artists'].str.contains('Various Artists', case=False, na=False)].copy()

print(f"\n✅ Après suppression : {len(df):,} tracks")
print(f"✅ Artistes uniques : {df['artists'].nunique():,}")
print(f"🗑️ Tracks supprimées : {various_count:,}")

print("\n✅ Prêt pour le filtrage TOP 3000!")
```

**Créer un dataset propre avec les 3000 meilleurs artistes.**

```python
print("=" * 80)
print("FILTRAGE : TOP 3000 ARTISTES")
print("=" * 80)

# Obtenir le top 3000 artistes
top_artists = df['artists'].value_counts().head(3000).index.tolist()

# Filtrer le dataset
df_filtered = df[df['artists'].isin(top_artists)].copy()

print(f"\n✅ Dataset original : {len(df):,} tracks")
print(f"✅ Dataset filtré : {len(df_filtered):,} tracks")
print(f"✅ Artistes conservés : {df_filtered['artists'].nunique():,}")
print(f"✅ Réduction : {(1 - len(df_filtered)/len(df))*100:.1f}%")
```

**Nettoyage des valeurs manquantes**

```python
print("=" * 80)
print("NETTOYAGE DES VALEURS MANQUANTES")
print("=" * 80)

print(f"\n📊 Avant nettoyage : {len(df_filtered):,} lignes")

# Supprimer les lignes avec name ou artists manquants
df_cleaned = df_filtered.dropna(subset=['name', 'artists'])

print(f"✅ Après nettoyage : {len(df_cleaned):,} lignes")
print(f"🗑️ Lignes supprimées : {len(df_filtered) - len(df_cleaned)}")

# Vérifier qu'il n'y a plus de manquants sur les colonnes critiques
print(f"\n✅ Valeurs manquantes restantes :")
missing = df_cleaned[['name', 'artists', 'album']].isnull().sum()
print(missing)
```

**Normalisation des noms**

```python
print("=" * 80)
print("NORMALISATION DES DONNÉES")
print("=" * 80)

# Nettoyer les espaces en trop
df_cleaned['artists'] = df_cleaned['artists'].str.strip()
df_cleaned['name'] = df_cleaned['name'].str.strip()
df_cleaned['album'] = df_cleaned['album'].str.strip()

# Afficher quelques exemples
print("\n📋 Exemples après nettoyage :")
print(df_cleaned[['artists', 'name', 'album']].head(10))
```

**Identifier les colonnes à garder**

```python
print("=" * 80)
print("SÉLECTION DES COLONNES")
print("=" * 80)

print("\n📋 Colonnes disponibles :")
for i, col in enumerate(df_cleaned.columns, 1):
    print(f"  {i}. {col}")

# Colonnes essentielles pour Neo4j
essential_cols = ['id', 'name', 'album', 'album_id', 'artists']

# Colonnes audio features (pour MongoDB)
audio_features = [col for col in df_cleaned.columns if col not in essential_cols]

print(f"\n✅ Colonnes essentielles (Neo4j) : {essential_cols}")
print(f"✅ Audio features (MongoDB) : {audio_features[:10]}...")  # Afficher les 10 premières
```

**Vérification finale et statistiques**

```python
print("=" * 80)
print("STATISTIQUES FINALES")
print("=" * 80)

print(f"\n📊 Dataset nettoyé :")
print(f"  - Tracks : {len(df_cleaned):,}")
print(f"  - Artistes uniques : {df_cleaned['artists'].nunique():,}")
print(f"  - Albums uniques : {df_cleaned['album'].nunique():,}")
print(f"  - Colonnes : {len(df_cleaned.columns)}")

print(f"\n🎤 Top 10 artistes du dataset nettoyé :")
top_10 = df_cleaned['artists'].value_counts().head(10)
for i, (artist, count) in enumerate(top_10.items(), 1):
    print(f"  {i}. {artist}: {count} tracks")

print(f"\n✅ Dataset prêt pour l'export !")
```

**Export du dataset nettoyé**

```python
print("=" * 80)
print("EXPORT DU DATASET NETTOYÉ")
print("=" * 80)

# Export en CSV (pour backup)
csv_path = '../data/processed/tracks_cleaned.csv'
df_cleaned.to_csv(csv_path, index=False)
print(f"✅ CSV exporté : {csv_path}")

# Export en XLSX (Excel)
xlsx_path = '../data/processed/tracks_cleaned.xlsx'
df_cleaned.to_excel(xlsx_path, index=False, engine='openpyxl')
print(f"✅ XLSX exporté : {xlsx_path}")
print(f"📊 Taille : {len(df_cleaned):,} lignes × {len(df_cleaned.columns)} colonnes")

# Export liste unique d'artistes (CSV et XLSX)
artists_df = df_cleaned[['artists']].drop_duplicates().sort_values('artists')
artists_df.to_csv('../data/processed/artists_unique.csv', index=False)
artists_df.to_excel('../data/processed/artists_unique.xlsx', index=False, engine='openpyxl')

print(f"\n✅ Liste d'artistes uniques exportée (CSV et XLSX)")
print(f"📊 Nombre d'artistes : {len(artists_df):,}")

print("\n" + "=" * 80)
print("✅ TOUS LES FICHIERS EXPORTÉS AVEC SUCCÈS !")
print("=" * 80)
print("\n📁 Fichiers créés dans data/processed/ :")
print("  1. tracks_cleaned.csv")
print("  2. tracks_cleaned.xlsx  ⭐")
print("  3. artists_unique.csv")
print("  4. artists_unique.xlsx")
```
